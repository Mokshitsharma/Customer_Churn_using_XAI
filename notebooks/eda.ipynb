```python
import pandas as pd
import numpy as np
import re
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("C:\\Users\\dell\\OneDrive\\Desktop\\xai-financial-sentiment\\data\\raw\\data.csv")
```

```python
df = df.rename(columns={
    "Sentence": "text",
    "Sentiment": "label"
})
```

```python
df.head()
```

```python
df.info()
```

```python
print(df["label"].value_counts())
```

```python
def clean_text(text):
    text = str(text)
    
    text = re.sub(r"\$\w+", "", text)  # remove stock tickers
    text = re.sub(r"[^a-zA-Z\s]", " ", text)
    text = re.sub(r"\s+", " ", text)
    return text.lower().strip()
```

```python
df["clean_text"] = df["text"].apply(clean_text)
```

```python
df.head()
```

```python
label_map = {
    "positive": 1,
    "neutral": 0,
    "negative": -1
}

df["target"] = df["label"].map(label_map)
```

```python
# remove empty or very short text
df = df[df["clean_text"].str.len() > 20]
```

```python
# remove nulls
df = df.dropna()
```

```python
df.shape
```

```python
df["target"].value_counts().plot(kind="bar", title="Label Distribution")
```

```python
df = df.reset_index(drop=True)
```

```python
df.to_csv("C:\\Users\\dell\\OneDrive\\Desktop\\xai-financial-sentiment\\data\\processed\\cleaned_data.csv", index=False)
```
